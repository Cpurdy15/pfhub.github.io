# Involvement

## How to Participate in the Benchmark Effort

All those interested in phase field modeling are welcome to participate in the effort and to join our discussions. We value all forms of productive input. There are a number of ways to get involved:

# Uploading Benchmark Results

The most significant way to contribute is by trying to implement the [benchmark problems][benchmark problems] using one of the [suggested codes][suggested codes] or your own custom code and then [uploading the results][results] to the website.

# Chat

We have active discussion on the [Gitter chat channel][gitter chat]. Feel free to ask questions there.

# Chat Meetings

We hold monthly [chat meetings on Gitter][gitter chat]. Please try and participate and either ask question or listen to the chat. The content from [previous chat meetings][chat meetings] is available.

# Mailing List

Please subscribe to the [mailing list][mailing list] to stay informed about upcoming chat meetings and workshops or just to ask questions.

# Via GitHub

Please feel free to [open an issue on GitHub][open issue github] if you would like to raise an issue about one of [the benchmarks problems][benchmark problems] or with anything at all about the website.

# Workshops

An integral component of PFHub is regular [workshops][workshop] and associated [hackathons][hackathon]. Approximately twice a year, a spectrum of United States and international researchers, ranging from graduate students to senior scientists, come together to participate. See the [workshops page][workshop] for details about past workshops and plans for future workshops.

# Website Development

The development of this website is a community effort. Please see the [development guide][development guide] if you would like to help out.
