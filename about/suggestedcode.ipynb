# Suggested Phase Field Codes
Codes with well documented, annonated examples that demonstrate the phase field method. See the [submission page](https://pages.nist.gov/pfhub/submit_a_new_code) to submit a new code to this list.




## Moose

A multiphysics finite element object oriented simulation environment written in C++.

[Fe-Cr Phase Decomposition](http://mooseframework.org/wiki/MooseTutorials/IronChromiumDecomposition/) | [Cahn-Hilliard](https://github.com/idaholab/moose/tree/devel/modules/phase_field/examples/cahn-hilliard) | [Grain Growth](https://github.com/idaholab/moose/tree/devel/modules/phase_field/examples/grain_growth) | [Nucleation](https://github.com/idaholab/moose/tree/devel/modules/phase_field/examples/nucleation)




## FEniCS

The FEniCS Project is a collection of free software with an extensive list of features for automated, efficient solution of differential equations written in C++.

[Cahn-Hilliard](http://fenicsproject.org/documentation/dolfin/1.1.0/python/demo/pde/cahn-hilliard/python/documentation.html)




## FiPy

A finite Volume PDE solver written in Python.

[Cahn-Hilliard](http://www.ctcms.nist.gov/fipy/examples/cahnHilliard/generated/examples.cahnHilliard.mesh2DCoupled.html) | [Quaternary](http://www.ctcms.nist.gov/fipy/examples/phase/generated/examples.phase.quaternary.html) | [Anisotropy](http://www.ctcms.nist.gov/fipy/examples/phase/generated/examples.phase.anisotropy.html) | [Polycrystal](http://www.ctcms.nist.gov/fipy/examples/phase/generated/examples.phase.polyxtalCoupled.html) | [Hackathon repo](https://github.com/guyer/phase_field_hackathon_I) | [Automated Benchmark 1](https://github.com/usnistgov/FiPy-spinodal-decomposition-benchmark/tree/master)
