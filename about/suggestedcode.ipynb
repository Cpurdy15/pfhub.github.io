Codes with well documented, annonated examples that demonstrate the phase field method. See the [submission page][submission page] to submit a new code to this list.

## The Codes

```python
import pandas as pd
from IPython.display import HTML, display
from html import escape

# Load the CSV file
df = pd.read_csv("codes.csv", dtype=str).fillna("")

# Store the HTML for all codes
all_codes_html = ""

# Loop through each code in the CSV
for index, code in df.iterrows():

    name = escape(code["name"])
    description = escape(code["description"])
    language = escape(code["language"])
    github = escape(code["github"], quote=True)
    examples = code["examples"]

    # Create clickable example links
    example_links = []

    for example in examples.split(";"):
        if "|" in example:
            title, url = example.split("|", 1)

            title = escape(title.strip())
            url = escape(url.strip(), quote=True)

            if url:
                example_links.append(
                    f'<a href="{url}">{title}</a>'
                )

    examples_html = " | ".join(example_links)

    # Create the GitHub link
    github_html = (
        f'<a href="{github}">GitHub Repository</a>'
        if github else ""
    )

    # Create the HTML card
    all_codes_html += f"""
    <div style="
        background-color: #f0f4df;
        padding: 20px;
        margin-bottom: 15px;
        border-radius: 5px;
        color: #333;
    ">

        <h3 style="margin-top: 0;">
            {name}
        </h3>

        <p>{description}</p>

        <p>
            <strong>Language:</strong> {language}
        </p>

        <p>{github_html}</p>

        <p>{examples_html}</p>

    </div>
    """

# Display all codes
display(HTML(all_codes_html))
```
